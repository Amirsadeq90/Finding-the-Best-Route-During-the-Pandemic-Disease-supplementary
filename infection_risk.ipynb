```python
import pandas as pd
import numpy as np
from bidi.algorithm import get_display
import math
```

```python
def av_dist(w,h):
    d = math.sqrt(w**2+h**2)
    return 1/15*(math.pow(w,3)/math.pow(h,2)+math.pow(h,3)/math.pow(w,2)
                +d*(3-math.pow(w,2)/math.pow(h,2)-math.pow(h,2)/math.pow(w,2))
                +5/2*(h**2/w*math.log((w+d)/h,10)+w**2/h*math.log((h+d)/w,10)))
```

```python
rho = 0.007682
```

```python
def f(type,z):
    if type == 'p':
        return 1-math.exp((-0.00143853*1740+0.71455401)*(rho*40)/(av_dist(4,30.3)**2)*z)
    elif type == 's':
        return 1-math.exp((-0.00180107*780+0.82402941)*(rho*180)/(av_dist(2.6,19.52)**2)*z)
    elif type == 'br':
        return 1-math.exp((-0.00149112*780+0.38875338)*(rho*150)/(av_dist(2.55,17.9)**2)*z)
    elif type == 'b':
        return 1-math.exp((-0.00220276*780+0.56565911)*(rho*80)/(av_dist(2.55,12)**2)*z)
    elif type == 'c':
        return 1-math.exp((-2.729480)*(rho*4)/(av_dist(1.2,1.5)**2)*z)
```

# Infection Risk Comparison Across Routes

```python
import matplotlib.pyplot as plt
import numpy as np

routes = ['R1', 'R2', 'R3', 'R4', 'R5', 'R6', 'R7', 'R8', 'R9', 'R10', 'R11']
risk_values = [R1,R2,R3,R4,R5,R6,R7,R8,R9,R10,R11] 

plt.figure(figsize=(10, 6))
bars = plt.bar(routes, risk_values, color='skyblue')
max_index = np.argmax(risk_values)
min_index = np.argmin(risk_values)
bars[max_index].set_color('red')
bars[min_index].set_color('green')

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.2f}',
             ha='center', va='bottom')

plt.title('Infection Risk Comparison Across Routes', fontsize=14)
plt.xlabel('Routes', fontsize=12)
plt.ylabel('Infection Risk Percentage', fontsize=12)
plt.ylim(0, max(risk_values) * 1.2) 
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig('Eng/InfectionRiskComparisonAcrossRoutes.jpg')
```

# Impact of Room Size on COVID-19 Transmission Risk

# Knee of pedestrian average distance figure

```python
import pandas as pd
import numpy as np
import math
from bidi.algorithm import get_display
from kneed import KneeLocator
import matplotlib.pyplot as plt
from matplotlib import rcParams

rcParams['figure.figsize'] = (12, 8)
rcParams['font.size'] = 12
rcParams['axes.grid'] = True
rcParams['grid.alpha'] = 0.3

def f(n, z, k, r):
    n = rho * n
    return 1 - math.exp(n * k * z / (r ** 2))

def av_dist(w, h):
    d = math.sqrt(w ** 2 + h ** 2)
    return (1/15 * (math.pow(w, 3)/math.pow(h, 2) + math.pow(h, 3)/math.pow(w, 2) +
            d * (3 - math.pow(w, 2)/math.pow(h, 2) - math.pow(h, 2)/math.pow(w, 2)) +
            5/2 * (h**2/w * math.log((w+d)/h, 10) + w**2/h * math.log((h+d)/w, 10))))
    
result = []
k = -0.00143853 * 1740 + 0.71455401
for i in range(1, 11):
    for j in range(1, 101):
        av_distance = av_dist(4, j)
        result.append((i, j, av_distance, f(i*j, 1, k, av_distance)))
df = pd.DataFrame(result, columns=['density', 'length', 'av_distance', 'probability'])

plt.figure(figsize=(14, 8))
ax = plt.gca()

markers = ['.', '*', ',', 'o', 'v', '>', '<', '1', '2', '3']
colors = plt.cm.viridis(np.linspace(0, 1, 10)) 
labels = [get_display(f'{0.25 + 0.25*i} person per square meter') for i in range(10)]

knee_points = []
for i in range(1, 11):
    subset = df[df['density'] == i]
    x = subset['length'].values
    y = subset['probability'].values
    
    # Find knee point
    kneedle = KneeLocator(x, y, curve='convex', direction='decreasing', online=True)
    if kneedle.knee is not None:
        knee_x, knee_y = kneedle.knee, kneedle.knee_y
        knee_points.append((knee_x, knee_y, i-1))
    
    # Plot curve
    ax.plot(x, y, 
            marker=markers[i-1], 
            markersize=8 if markers[i-1] not in [',', '.'] else 4,
            markevery=5,
            lw=2,
            color=colors[i-1],
            alpha=0.8,
            label=labels[i-1])

for knee_x, knee_y, idx in knee_points:
    ax.plot(knee_x, knee_y, 'ro', markersize=10, markeredgecolor='k')
    ax.text(knee_x, knee_y + 0.05, f'Knee: {knee_x:.1f}m', 
            ha='center', va='bottom', fontsize=10,
            bbox=dict(facecolor='white', alpha=0.8, edgecolor='none'))

reshaped_x_label = 'The length of the sidewalk'
x_label = get_display(reshaped_x_label)
reshaped_y_label = 'The probability of contracting disease during one hour of walking'
y_label = get_display(reshaped_y_label)
ax.set_xlabel(x_label, fontsize=14)
ax.set_ylabel(y_label, fontsize=14)
ax.set_title('Disease Transmission Probability vs. Sidewalk Length', fontsize=16, pad=20)
legend = ax.legend(loc='upper left', bbox_to_anchor=(1, 1), framealpha=1)
legend.set_title('Population Density', prop={'size': 12})
plt.tight_layout()
plt.savefig('Eng/SideWalk_Length_Improved.jpg', dpi=300, bbox_inches='tight')
plt.show()
```

# Infection Spread Simulation

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from matplotlib.colors import LinearSegmentedColormap
import math

m = 10  # rows
l = 15  # columns
n = 5   # number of infected individuals
sample_cell = 7

grid = np.zeros((m, l))
infected_positions = []
for _ in range(n):
    while True:
        pos = (np.random.randint(0, m), np.random.randint(0, l))
        if pos not in infected_positions:
            infected_positions.append(pos)
            grid[pos] = 1  # 1 represents infected
            break
            
start_row = np.random.randint(0, m)
path = [(start_row, col) for col in range(l)]
durations = np.random.randint(1, 5, size=l)

def phi(r):
    return 1/(1 + r**2) 

fig, ax = plt.subplots(figsize=(16, 10))
cmap = LinearSegmentedColormap.from_list('infection', ['green', 'yellow', 'red'])
for i in range(m):
    for j in range(l):
        rect = patches.Rectangle((j, m-1-i), 1, 1, linewidth=1, 
                               edgecolor='black', facecolor='white')
        ax.add_patch(rect)

for pos in infected_positions:
    i, j = pos
    circ = patches.Circle((j + 0.5, m-1-i + 0.5), 0.4, 
                         facecolor='red', edgecolor='black')
    ax.add_patch(circ)
    ax.text(j + 0.5, m-1-i + 0.5, 'INF', 
           ha='center', va='center', color='white', fontsize=10, weight='bold')

for step in range(l):
    pos = path[step]
    x, y = pos
    T_j = durations[step]
    distances = []
    probabilities = []
    for inf_pos in infected_positions:
        xi, yi = inf_pos
        distance = math.sqrt((xi - x)**2 + (yi - y)**2)
        distances.append(distance)
        probabilities.append(phi(distance))
        
    total_prob = 1 - np.prod([1 - p for p in probabilities])
    rect = patches.Rectangle((y, m-1-x), 1, 1, linewidth=2, 
                           edgecolor='blue' if step == sample_cell else 'blue', 
                           facecolor=cmap(total_prob), 
                           alpha=0.7 if step != sample_cell else 1.0)
    ax.add_patch(rect)
    if step == sample_cell:
        ax.text(y + 0.5, m-1-x + 0.8, f'Cell {step+1}\nT={T_j}', 
               ha='center', va='center', fontsize=10, weight='bold',
               bbox=dict(facecolor='white', alpha=0.9, edgecolor='black'))
        detail_text = "Detailed Variables:\n"
        for idx, (dist, prob) in enumerate(zip(distances, probabilities)):
            detail_text += f"r{idx+1} = {dist:.2f}\nϕ(r{idx+1}) = {prob:.3f}\n"
        detail_text += f"\nTotal ϕ = {total_prob:.3f}"
        ax.text(l + 1.5, m/2, detail_text, 
               ha='left', va='center', fontsize=10,
               bbox=dict(facecolor='white', alpha=0.9, edgecolor='black', pad=10))
        for idx, inf_pos in enumerate(infected_positions):
            xi, yi = inf_pos
            ax.plot([y + 0.5, yi + 0.5], [m-1-x + 0.5, m-1-xi + 0.5], 
                   'k--', alpha=0.5, linewidth=1)
            ax.text((y + yi)/2 + 0.5, (m-1-x + m-1-xi)/2 + 0.5, 
                   f'r{idx+1}', fontsize=8, 
                   bbox=dict(facecolor='white', alpha=0.8, edgecolor='none'))
    else:
        ax.text(y + 0.5, m-1-x + 0.5, f'{step+1}', 
               ha='center', va='center', fontsize=8)

sample_x, sample_y = path[sample_cell]
ax.scatter(sample_y + 0.5, m-1-sample_x + 0.5, s=300, 
          marker='*', color='gold', edgecolor='black', zorder=10)

for step in range(l-1):
    start = path[step]
    end = path[step+1]
    ax.arrow(start[1] + 0.5, m-1-start[0] + 0.5, 
             end[1] - start[1], start[0] - end[0], 
             head_width=0.2, head_length=0.2, fc='blue', ec='blue', lw=2)

start_x, start_y = path[0]
ax.text(start_y + 0.5, m-1-start_x + 0.8, 'START', 
       ha='center', va='center', color='blue', fontsize=12, weight='bold')
end_x, end_y = path[-1]
ax.text(end_y + 0.5, m-1-end_x + 0.8, 'END', 
       ha='center', va='center', color='blue', fontsize=12, weight='bold')

sm = plt.cm.ScalarMappable(cmap=cmap, norm=plt.Normalize(vmin=0, vmax=1))
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax, orientation='vertical', fraction=0.046, pad=0.04)
cbar.set_label('Infection Probability $\phi(r_{ij})$', fontsize=12)

ax.set_title('Infection Spread Simulation\n(Highlighting detailed variables for one sample cell and sample $\phi(r_{ij})$)', fontsize=14)
ax.set_xlabel('Column', fontsize=12)
ax.set_ylabel('Row', fontsize=12)
ax.set_xlim(-0.5, l + 5) 
ax.set_xticks(np.arange(0, l+1, 1))
ax.set_yticks(np.arange(0, m+1, 1))
ax.grid(True, alpha=0.3)
ax.set_aspect('equal')

plt.tight_layout()
plt.savefig('Eng/Infection_Spread_Simulation.jpg', dpi=300, bbox_inches='tight')
plt.show()
```
